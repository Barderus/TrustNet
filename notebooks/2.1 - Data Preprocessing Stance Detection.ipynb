## Build the model-ready data

The notebook writes two CSVs under data/preprocessed/: master_stance.csv and stance_preprocessed.csv. It does not create evaluation reports or run folders.


```python
import pandas as pd

from utils.preprocessing import build_master_stance, build_stance_preprocessed

MASTER_PATH = "data/preprocessed/master_stance.csv"
MODEL_PATH = "data/preprocessed/stance_preprocessed.csv"
```

```python
master = build_master_stance()
model_ready = build_stance_preprocessed(master)

master.to_csv(MASTER_PATH, index=False)
model_ready.to_csv(MODEL_PATH, index=False)
print("Saved", MASTER_PATH, "and", MODEL_PATH)
print("Rows:", len(master), "master;", len(model_ready), "model-ready")
```
