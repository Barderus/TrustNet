## Classical Model Comparison

Macro F1 is important for stance detection because the four stance classes are not equally common.

````python
if latest_run is None:
    print("No stance baseline results found yet. Run script/run_classical_baselines.py first.")
else:
    metrics = pd.read_csv(latest_run / "metrics.csv")
    columns = [
        "model",
        "accuracy",
        "macro_precision",
        "macro_recall",
        "macro_f1",
        "weighted_f1",
        "roc_auc",
    ]
    comparison = metrics[[column for column in columns if column in metrics.columns]]
    comparison = comparison.sort_values("macro_f1", ascending=False)
    display(comparison)
````

Use the table above to identify the best classical stance model. Pay close attention to macro F1 because a model can look good on accuracy while still doing poorly on smaller classes such as disagree.
