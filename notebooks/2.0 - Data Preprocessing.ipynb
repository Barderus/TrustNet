## Build the model-ready data

The notebook writes two CSVs under data/preprocessed/: master_fakenews.csv and fakenews_preprocessed.csv. It does not create evaluation reports or run folders.


```python
import pandas as pd

from utils.preprocessing import build_master_fakenews, build_fakenews_preprocessed

MASTER_PATH = "data/preprocessed/master_fakenews.csv"
MODEL_PATH = "data/preprocessed/fakenews_preprocessed.csv"
```

```python
master = build_master_fakenews()
model_ready = build_fakenews_preprocessed(master)

master.to_csv(MASTER_PATH, index=False)
model_ready.to_csv(MODEL_PATH, index=False)
print("Saved", MASTER_PATH, "and", MODEL_PATH)
print("Rows:", len(master), "master;", len(model_ready), "model-ready")
```
